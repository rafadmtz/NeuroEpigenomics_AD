```python
import polars as pl

SUBCLASSES= ["Vip", "L2-3-IT", "Sst"]

TEs= ["LINE", "LTR", "SINE"]

vip_da= pd.read_csv("../results/06_differential_accessibility/da_Vip_trend.csv.gz", index_col="peak_id")

vipda_te = (
    vip_da.merge(peak_overlap, left_on="peak_id", right_index=True, how="left")
)


vipda_tef=vipda_te[
    (vipda_te["repClass"].isin(TEs)) & (vipda_te["padj"]<= 0.05)
    ]



vipda_tef.head()
```
